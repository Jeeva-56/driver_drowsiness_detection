```python
import numpy as np
import matplotlib.pyplot as plt

# Image Preprocessing and Organizing
from PIL import Image
import os
import random
import shutil
```

# Check Whether Any Images Are Corrupted

```python
# Image Folder Path|

path = "../datasets/images"
```

### Read And Check Whether Images Are Corrupted

```python
def check_corrupted_images(path):

    corrupted_images = {}

    for class_name in os.listdir(path):

        class_path = os.path.join(path, class_name)
        
        corrupted_images[class_name] = []

        for file in os.listdir(class_path):

            image_path = os.path.join(class_path, file)

            try:
                image = Image.open(image_path)
                image.verify()
                
            except:
                corrupted_images[class_name].append(image_path)

    return corrupted_images
```

```python
corrupted_images = check_corrupted_images(path)

for class_name, images in corrupted_images.items():
    
    print(f"{class_name:7}: {len(images)}")
```

# Splitting Images - Train, Test, Validation

```python
source_path = "../datasets/images"
destination_path = "../datasets"

classes = os.listdir(source_path)

# Split Ratio
train_ratio = 0.70
validation_ratio = 0.15
test_ratio = 0.15
```

```python
for class_name in classes:

    # Get Class Path
    class_path = os.path.join(source_path, class_name)

    # Get All Images
    images = os.listdir(class_path)

    # Shuffle Images
    random.shuffle(images)

    # Calculate Split Index
    total_images = len(images)

    train_index = int(total_images * train_ratio)
    validation_index = int(total_images * validation_ratio) + train_index


    # Split Images into Train, Validation, and Test Sets
    train_data = images[:train_index]
    validation_data = images[train_index:validation_index]
    test_data = images[validation_index:]

  
    # Split Eye and Mouth Datasets
    if class_name in ['closed', 'open']:
        dataset_name = "eye_dataset"
    else:
        dataset_name = "mouth_dataset"


    # Create Train, Validation, and Test Folders
    splits = ["train", "validation", "test"]
       
    for split in splits:

        split_path = os.path.join(destination_path, split, dataset_name, class_name)

        os.makedirs(split_path, exist_ok=True)


    # Copy Training Images
    for image in train_data:

        source_image_path = os.path.join(class_path, image)

        destination_image_path = os.path.join(
            destination_path, splits[0], dataset_name, class_name, image
        )

        shutil.copy(source_image_path, destination_image_path)


    # Copy Validation Images
    for image in validation_data:

        source_image_path = os.path.join(class_path, image)

        destination_image_path = os.path.join(
            destination_path, splits[1], dataset_name, class_name, image
        )

        shutil.copy(source_image_path, destination_image_path)


    # Copy Test Images
    for image in test_data:

        source_image_path = os.path.join(class_path, image)

        destination_image_path = os.path.join(
            destination_path, splits[2], dataset_name, class_name, image
        )

        shutil.copy(source_image_path, destination_image_path)


    print("-" * 50)
    print(class_name)
    print(
        f"Train = {len(train_data)} | "
        f"Validation = {len(validation_data)} | "
        f"Test = {len(test_data)}"
    )
```
